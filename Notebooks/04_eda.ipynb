Columnas del dataset:

●	age: La edad del cliente.

●	marital: El estado civil del cliente.

●	education: El nivel educativo del cliente.

●	campaign: El número de contactos realizados durante esta campaña para este cliente.

●	pdays: Número de días que han pasado desde la última vez que se contactó con el cliente durante esta campaña.

●	previous: Número de veces que se ha contactado con el cliente antes de esta campaña.

●	poutcome: Resultado de la campaña de marketing anterior.

●	emp.var.rate: La tasa de variación del empleo.

●	cons.price.idx: El índice de precios al consumidor.

●	cons.conf.idx: El índice de confianza del consumidor.

●	euribor3m: La tasa de interés de referencia a tres meses.

●	nr.employed: El número de empleados.

●	y: Indica si el cliente ha suscrito un producto o servicio (Sí/No).

●	date: La fecha en la que se realizó la interacción con el cliente.

●	ID: Identificador único del cliente.

●	y_num: Indica si el cliente ha suscrito un producto o servicio en formato numérico (1: Sí, 0: No).

●	contact_month: Mes en el que se realizó la interacción con el cliente durante la campaña de marketing.

●	contact_year: Año en el que se realizó la interacción con el cliente durante la campaña de marketing.

●	Income: Representa el ingreso anual del cliente en términos monetarios.

●	Kidhome: Indica el número de niños en el hogar del cliente.

●	Teenhome: Indica el número de adolescentes en el hogar del cliente.

●	Dt_Customer: Representa la fecha en que el cliente se convirtió en cliente de la empresa.

●	NumWebVisitsMonth: Indica la cantidad de visitas mensuales del cliente al sitio web de la empresa.

##### Factores demográficos y financieros

Pregunta clave 1:

¿Qué características demográficas y financieras de los clientes están más asociadas a la suscripción del depósito a plazo?

- Las categorías que muestran mayor tasa de suscripción: estudiantes y jubilados, solteros, sin estudios

- La presencia de préstamos hipotecarios activos aumenta un poco la probabilidad de suscripción.
Pero otros tipos de préstamos la reducen.

- La edad media de los suscriptores es aprox. 39 años. El grupo de éxito (yes) muestra una dispersión de edad ligeramente superior.

- El nivel de ingresos de cliente no parece ser un factor determinante para suscripción.

```python
# Tasas de suscripción por categoría
cat_vars = ["job", "marital", "education", "loan", "housing"]

for col in cat_vars:
    rate = df_eda.groupby(col)['y'].value_counts(normalize=True).unstack()
    rate['yes'].sort_values(ascending=False).plot(kind="bar")
    plt.title(f"Tasa de suscripción por *{col}*")
    plt.ylabel("Proporción de *yes*")
    plt.show()
```

```python
# Edad y ingresos por Boxplots
for col in ['age', 'Income']:
    plt.figure(figsize=(8,4))
    sns.boxplot(x='y', y=col, data=df_eda)
    plt.title(f'{col} vs Suscripción')
    plt.show()
```

##### Canal y momento temporal

Pregunta clave 3:

¿Existen diferencias en la efectividad de la campaña según el canal de contacto y el momento temporal (mes/año)?

- La efectividad de la campaña varía según el canal de contacto, canal 'celular' genera mayor tasa de suscripción que 'telefono'.

- La tasa de suscripción tiene el máximo en 2016 y una caída en 2017, seguida de una ligera recuperación en 2018.

- Hay la estacionalidad de la tasa de suscripción a lo largo del año: pico máximo en Octubre y punto más bajo en Septiembre.
La tasa no es estable, presenta fluctuaciones constantes con una recuperación notable en el último trimestre del año.

```python
# Tasa de éxito por canal
rate_contact = df_eda.groupby("contact")["y"].apply(lambda x: (x == "yes").mean())

rate_contact.plot(kind="bar")
plt.title("Tasa de suscripción por canal de contacto")
plt.ylabel("Proporción de yes")
plt.show()
```

```python
# Análisis temporal por año
df_year = df_eda.loc[:, ["contact_year", "y"]]

rate_year = df_year.groupby("contact_year")["y"].apply(lambda x: (x == "yes").mean())
rate_year.plot(marker="o")

plt.title("Tasa de suscripción por año")
plt.ylabel("Proporción de yes")
plt.xlabel("Año")
plt.show()
```

```python
# Análisis temporal por mes
df_month = df_eda.loc[:, ["contact_month", "y"]]

rate_month = df_month.groupby("contact_month")["y"].apply(lambda x: (x == "yes").mean())
rate_month.plot(marker="o")

plt.title("Tasa de suscripción por mes")
plt.ylabel("Proporción de yes")
plt.xlabel("Mes")
plt.show()
```

##### Historial de campañas previas
Pregunta clave 4:

¿El historial de contactos y campañas previas mejora o reduce la tasa de suscripción del producto?

- Los clientes con contactos previos exitosos presentan la mayor probabilidad de suscripción.

- Un mayor número de contactos previos no siempre mejora el resultado, sugiriendo posible saturación.

- La experiencia de las campañas previas influye en la decisión positivamente.

```python
# Calculamos la cuota de suscripción en función del número de contactos previos
df_eda.groupby("previous")["y"].apply(lambda x: (x == "yes").mean())
# Incrementar contactos sin éxito previo no garantiza mejores resultados.
```

```python
# Días desde el último contacto

# Creamos una variable binaria y contamos la cuota de suscripción:
df_eda["contactado_antes"] = df_eda["pdays"] != 999
rate = df_eda.groupby("contactado_antes")["y"].apply(lambda x: (x == "yes").mean())
rate
```

```python
# Vamos a trazar un gráfico
rate.plot(kind="bar")
plt.title("Influencia del contacto previo en la suscripción")
plt.ylabel("Proporción de yes")
plt.xticks([0,1], ["No contactado", "Contactado antes"], rotation=0)
plt.show()
```

```python
# Resultado de la campaña previa
rate_poutcome = df_eda.groupby("poutcome")["y"].apply(lambda x: (x == "yes").mean())

rate_poutcome.plot(kind="bar")
plt.title("Tasa de suscripción según resultado previo")
plt.ylabel("Proporción de yes")
plt.show()
```

##### Contexto macroeconómico

Pregunta clave 5:

¿Cómo influyen las condiciones macroeconómicas en la efectividad de la campaña y en la probabilidad de suscripción del depósito?

- Hay una correlación negativa clara entre el número de empleados y el éxito de la campaña.
Los productos se venden mucho mejor cuando la economía está en incertidumbre o enfriamiento.
Son productos que la gente busca por seguridad o ahorro cuando las cosas van peor.

```python
# Correlación entre variables macroeconómicas:
df_eda[['nr.employed','euribor3m','emp.var.rate']].corr()
# Las tres variables macroeconómicas están presentan correlaciones positivas muy altas entre sí, 
# lo que indica que describen un mismo contexto económico.
```

```python
# Comparamos tasas de suscripción por uno de los indicadores macroeconómicos: nivel de empleo
prop_yes = df_eda.groupby('nr.employed')['y'].value_counts(normalize=True).unstack()['yes']

plt.figure(figsize=(10, 5))
sns.lineplot(x=prop_yes.index, y=prop_yes.values, marker='o')

plt.title('Evolución de la Tasa de Suscripción vs nr.employed')
plt.xlabel('Número de Empleados')
plt.ylabel('Proporción de "yes"')

plt.show()
```
