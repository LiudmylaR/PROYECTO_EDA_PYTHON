##### Cargamos el conjunto de datos desde un archivo XLSX (detalles de consumidores)

```python
# El archivo tiene 3 hojas, las cargamos una por una
df_2012 = pd.read_excel('../data/raw/customer-details.xlsx', sheet_name = "2012", index_col=0)
df_2013 = pd.read_excel('../data/raw/customer-details.xlsx', sheet_name = "2013", index_col=0)
df_2014 = pd.read_excel('../data/raw/customer-details.xlsx', sheet_name = "2014", index_col=0)
```

##### 

```python
# Concatenamos tres hojas XSLX en un único DataFrame
df_customers = pd.concat([df_2012, df_2013, df_2014])
```

## Analisis preliminar df_customers

```python
se.analisis_rapido(df_customers)
```

El dataset df_customers cuenta con 
43170 filas, donde cada una representa información demográfica y de comportamiento de compra de los clientes del banco y 6 columnas, que incluyen cuatro variables numéricas, una columna de fecha y una categórica.

Columnas del dataset:

●	Income: Representa el ingreso anual del cliente en términos monetarios.

●	Kidhome: Indica el número de niños en el hogar del cliente.

●	Teenhome: Indica el número de adolescentes en el hogar del cliente.

●	Dt_Customer: Representa la fecha en que el cliente se convirtió en cliente de la empresa.

●	NumWebVisitsMonth: Indica la cantidad de visitas mensuales del cliente al sitio web de la empresa.

●	ID: Identificador único del cliente.

El conjunto de datos **NO** tiene ni duplicados ni valores nulos.

##### Columnas numéricas:

```python
se.descriptivo_num(df_customers)
```

##### Descripción de los datos numéricos

Observación preliminar de los datos numericos:

El ingreso anual presenta una media cercana a 93.000, con una alta dispersión, lo que indica una gran variabilidad entre clientes.

La mayoría de los hogares tiene uno o dos niños o adolescentes, con valores medios cercanos a 1.

La variable Dt_Customer recoge la fecha de alta de los clientes, con registros comprendidos entre 01.01.2012 y 31.12.2014, lo que permite realizar análisis temporales de antigüedad.

El número de visitas mensuales al sitio web es relativamente alto, con una media de alrededor de 17 visitas, aunque se observan diferencias importantes entre clientes.
